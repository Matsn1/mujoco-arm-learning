```python
XML = """
<mujoco>
  <option timestep="0.01"/>                       <!-- physics runs at 100 Hz -->
  <worldbody>
    <light pos="0 0 3"/>
    <geom type="plane" size="1 1 .1" rgba=".9 .9 .9 1"/>               <!-- floor -->
    <camera name="top" pos="0 0 1.5" xyaxes="1 0 0 0 1 0"/>           <!-- camera looking down -->

    <body name="link1" pos="0 0 .05">                                  <!-- first arm link -->
      <joint name="j1" type="hinge" axis="0 0 1" damping="0.5"/>       <!-- shoulder rotates around z -->
      <geom type="capsule" fromto="0 0 0 .3 0 0" size=".02" rgba=".2 .4 .8 1"/>
      <body name="link2" pos=".3 0 0">                                 <!-- second link, at end of first -->
        <joint name="j2" type="hinge" axis="0 0 1" damping="0.5"/>     <!-- elbow -->
        <geom type="capsule" fromto="0 0 0 .25 0 0" size=".02" rgba=".2 .8 .4 1"/>
        <site name="tip" pos=".25 0 0" size=".02"/>                     <!-- the hand -->
      </body>
    </body>

    <body name="target" mocap="true" pos=".3 .2 .05">                  <!-- red ball we can move -->
      <geom type="sphere" size=".03" rgba="1 0 0 1" contype="0" conaffinity="0"/>
    </body>
  </worldbody>

  <actuator>                                       <!-- 2 motors, torque limited to ±2 -->
    <motor joint="j1" ctrlrange="-2 2"/>
    <motor joint="j2" ctrlrange="-2 2"/>
  </actuator>

  <sensor>                                         <!-- like encoders on a real robot -->
    <jointpos joint="j1"/> <jointpos joint="j2"/>
    <jointvel joint="j1"/> <jointvel joint="j2"/>
  </sensor>
</mujoco>
"""
model = mujoco.MjModel.from_xml_string(XML)
data = mujoco.MjData(model)

# apply constant torque for 1 second and watch it move
renderer = mujoco.Renderer(model, 240, 240)
frames = []
for i in range(100):
    data.ctrl[:] = [0.5, -0.3]
    mujoco.mj_step(model, data)
    if i % 4 == 0:
        renderer.update_scene(data, camera="top")
        frames.append(renderer.render())

print("joint sensors:", data.sensordata)
print("hand position:", data.site("tip").xpos)
media.show_video(frames, fps=25)
```

```python
L1, L2 = 0.30, 0.25   # link lengths. If you changed link 2 in the XML, change it here too

class ArmEnv:
    def __init__(self, seed=0):
        self.model = mujoco.MjModel.from_xml_string(XML)
        self.data = mujoco.MjData(self.model)
        self.rng = np.random.default_rng(seed)

    def reset(self):
        mujoco.mj_resetData(self.model, self.data)
        self.data.qpos[:] = self.rng.uniform(-np.pi, np.pi, 2)          # random start pose
        r = self.rng.uniform(0.15, 0.50); a = self.rng.uniform(-np.pi, np.pi)
        self.goal = np.array([r*np.cos(a), r*np.sin(a), 0.05])          # random reachable target
        self.data.mocap_pos[0] = self.goal                               # move the red ball there
        mujoco.mj_forward(self.model, self.data)
        self.t = 0
        return self.obs()

    def obs(self):
        q, qd = self.data.sensordata[:2], self.data.sensordata[2:4]     # joint angles, velocities
        return np.concatenate([np.cos(q), np.sin(q), qd, self.goal[:2]])

    def step(self, action):
        self.data.ctrl[:] = np.clip(action, -2, 2)
        for _ in range(2): mujoco.mj_step(self.model, self.data)        # control at 50 Hz
        self.t += 1
        dist = np.linalg.norm(self.data.site("tip").xpos[:2] - self.goal[:2])
        return self.obs(), -dist, dist < 0.02, self.t >= 100, dist
```

```python
def ik(x, y):
    c2 = np.clip((x*x + y*y - L1*L1 - L2*L2) / (2*L1*L2), -1, 1)
    q2 = np.arccos(c2)                                                   # elbow angle
    q1 = np.arctan2(y, x) - np.arctan2(L2*np.sin(q2), L1 + L2*np.cos(q2))  # shoulder angle
    return np.array([q1, q2])

def expert(env, kp=8.0, kd=1.0):
    q, qd = env.data.sensordata[:2], env.data.sensordata[2:4]
    err = ik(*env.goal[:2]) - q
    err = (err + np.pi) % (2*np.pi) - np.pi                              # take the short way around
    return kp*err - kd*qd

# test it on 100 random targets
env = ArmEnv(); succ = 0; steps = []
for ep in range(100):
    env.reset()
    for t in range(100):
        _, _, done, trunc, dist = env.step(expert(env))
        if done: succ += 1; steps.append(t); break
print("success:", succ, "/100   average steps:", np.mean(steps))
```

```python
import torch, torch.nn as nn

def collect(n_demos, seed):
    env = ArmEnv(seed); X, Y = [], []
    for ep in range(n_demos):
        obs = env.reset()
        for t in range(100):
            a = np.clip(expert(env), -2, 2)
            X.append(obs); Y.append(a)              # input: what the robot senses, label: what the expert did
            obs, _, done, trunc, _ = env.step(a)
            if done or trunc: break
    return np.array(X, dtype=np.float32), np.array(Y, dtype=np.float32)
```

```python
def train(X, Y, epochs=200):
    torch.manual_seed(0)
    net = nn.Sequential(nn.Linear(8,128), nn.ReLU(), nn.Linear(128,128), nn.ReLU(), nn.Linear(128,2))
    opt = torch.optim.Adam(net.parameters(), 1e-3)
    X, Y = torch.tensor(X), torch.tensor(Y) / 2.0       # scale torques to [-1, 1]
    for e in range(epochs):
        perm = torch.randperm(len(X))
        for i in range(0, len(X), 256):
            b = perm[i:i+256]
            loss = ((net(X[b]) - Y[b])**2).mean()        # copy the expert: mean squared error
            opt.zero_grad(); loss.backward(); opt.step()
    return net, loss.item()

def evaluate(net, n=100):
    env = ArmEnv(seed=999); s = 0                         # new targets the network never saw
    for ep in range(n):
        obs = env.reset()
        for t in range(100):
            with torch.no_grad():
                a = net(torch.tensor(obs, dtype=torch.float32)).numpy() * 2
            obs, _, done, trunc, _ = env.step(a)
            if done: s += 1; break
    return s / n

for n in [10, 50, 200]:
    X, Y = collect(n, seed=1)
    net, loss = train(X, Y)
    print(f"{n:>3} demos ({len(X)} samples)  training loss={loss:.4f}  success={evaluate(net):.0%}")
```

```python
def dagger(n_start=50, rounds=4, n_new=50):
    X, Y = collect(n_start, seed=1)
    net, _ = train(X, Y)
    results = [evaluate(net)]
    for r in range(rounds):
        env = ArmEnv(seed=100 + r); Xn, Yn = [], []
        for ep in range(n_new):
            obs = env.reset()
            for t in range(100):
                Xn.append(obs); Yn.append(np.clip(expert(env), -2, 2))  # expert labels the state...
                with torch.no_grad():
                    a = net(torch.tensor(obs, dtype=torch.float32)).numpy() * 2  # ...but the network drives
                obs, _, done, trunc, _ = env.step(a)
                if done or trunc: break
        X = np.concatenate([X, np.array(Xn, np.float32)])
        Y = np.concatenate([Y, np.array(Yn, np.float32)])
        net, _ = train(X, Y)
        results.append(evaluate(net))
        print(f"round {r+1}: {len(X)} samples, success = {results[-1]:.0%}")
    return net, results

dagger_net, dagger_results = dagger()
```

```python
import gymnasium as gym
from gymnasium import spaces
from stable_baselines3 import SAC

class ArmGym(gym.Env):
    def __init__(self, reward="dense", seed=0):
        self.arm = ArmEnv(seed); self.reward = reward
        self.observation_space = spaces.Box(-np.inf, np.inf, (8,), np.float32)
        self.action_space = spaces.Box(-1, 1, (2,), np.float32)      # RL works best with actions in [-1, 1]
    def reset(self, seed=None, options=None):
        return self.arm.reset().astype(np.float32), {}
    def step(self, action):
        obs, _, done, trunc, dist = self.arm.step(2 * action)          # scale back to torque ±2
        if self.reward == "dense":
            r = -dist                                                  # every step: how far am I?
        else:
            r = float(done) - 1.0                                      # sparse: -1 until success, nothing in between
        return obs.astype(np.float32), r, done, trunc, {}

def evaluate_rl(model, n=100):
    env = ArmGym(seed=999); s = 0
    for _ in range(n):
        o, _ = env.reset()
        while True:
            o, r, d, t, _ = env.step(model.predict(o, deterministic=True)[0])
            if d or t: s += d; break
    return s / n

model = SAC("MlpPolicy", ArmGym("dense"), seed=0)
rl_curve = []
for k in range(8):                       # 8 x 5000 = 40,000 steps
    model.learn(5000, reset_num_timesteps=False)
    rl_curve.append(evaluate_rl(model, 50))
    print(f"{(k+1)*5000} steps: success = {rl_curve[-1]:.0%}")
```

```python
class ArmGymErr(ArmGym):
    """Same task, but the policy also sees the error vector (target - hand), like a controller would."""
    def __init__(self, *a, **k):
        super().__init__(*a, **k)
        self.observation_space = spaces.Box(-np.inf, np.inf, (10,), np.float32)
    def _obs(self, obs):
        tip = self.arm.data.site("tip").xpos[:2]
        return np.concatenate([obs, self.arm.goal[:2] - tip]).astype(np.float32)
    def reset(self, seed=None, options=None):
        return self._obs(self.arm.reset()), {}
    def step(self, action):
        obs, _, done, trunc, dist = self.arm.step(2 * action)
        return self._obs(obs), -dist, done, trunc, {}

def evaluate_rl2(model, env_cls, n=50):
    env = env_cls(seed=999); s = 0; dists = []
    for _ in range(n):
        o, _ = env.reset()
        while True:
            o, r, d, t, _ = env.step(model.predict(o, deterministic=True)[0])
            if d or t:
                s += d; dists.append(np.linalg.norm(env.arm.data.site("tip").xpos[:2] - env.arm.goal[:2])); break
    return s / n, np.mean(dists)

model2 = SAC("MlpPolicy", ArmGymErr("dense"), seed=0)
rl_curve2 = []
for k in range(8):
    model2.learn(5000, reset_num_timesteps=False)
    succ, dist = evaluate_rl2(model2, ArmGymErr)
    rl_curve2.append(succ)
    print(f"{(k+1)*5000} steps: success = {succ:.0%}   avg final distance = {dist*100:.1f} cm")
```

```python
W = H = 480
def locate_target(env, renderer):
    renderer.update_scene(env.data, camera="top")
    img = renderer.render()
    mask = (img[:,:,0] > 150) & (img[:,:,1] < 80) & (img[:,:,2] < 80)     # red pixels
    if mask.sum() < 5:
        return None, img                                                   # ball hidden (e.g. under the arm)
    v, u = np.nonzero(mask)
    u, v = u.mean() + 0.5, v.mean() + 0.5                                  # centre of the red blob (pixels)

    cid = env.model.camera("top").id
    f = (H/2) / np.tan(np.deg2rad(env.model.cam_fovy[cid]) / 2)           # intrinsics: focal length in pixels
    p = env.data.cam_xpos[cid]                                             # camera position in world
    R = env.data.cam_xmat[cid].reshape(3, 3)                               # camera -> world rotation
    ray_cam = np.array([(u - W/2)/f, -(v - H/2)/f, -1.0])                  # MuJoCo camera looks down its -z
    ray_world = R @ ray_cam                                                # frame transform
    t = (0.05 - p[2]) / ray_world[2]                                       # hit the plane where the ball sits
    return (p + t*ray_world)[:2], img

env = ArmEnv(seed=3); env.reset()
renderer = mujoco.Renderer(env.model, H, W)
est, img = locate_target(env, renderer)
print("true :", env.goal[:2].round(4))
print("camera estimate:", None if est is None else est.round(4))
if est is not None: print("error:", round(np.linalg.norm(est - env.goal[:2])*1000, 1), "mm")
media.show_image(img)
```

```python
errors, succ, hidden = [], 0, 0
env = ArmEnv(seed=999)
for ep in range(100):
    obs = env.reset()
    est, _ = locate_target(env, renderer)
    if est is None: hidden += 1; continue
    errors.append(np.linalg.norm(est - env.goal[:2]))
    for t in range(100):
        obs[6:8] = est                                   # replace the true target with the camera's estimate
        with torch.no_grad():
            a = dagger_net(torch.tensor(obs, dtype=torch.float32)).numpy() * 2
        obs, _, done, trunc, _ = env.step(a)
        if done: succ += 1; break
print(f"hidden by arm: {hidden}   mean localisation error: {np.mean(errors)*1000:.1f} mm")
print(f"success using the camera: {succ}/{100-hidden}")
```
